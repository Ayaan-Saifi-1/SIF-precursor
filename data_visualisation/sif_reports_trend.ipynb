```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from IPython.display import IFrame
```

```python
# yahan apna actual data wire karna hai — bas ye function ya df replace kardo
# expected format:
# df_m: columns 'Period', 'Total_Reports', 'SIF_Potential' (Monthly: Jan..Dec)
# df_q: columns 'Period', 'Total_Reports', 'SIF_Potential' (Quarterly: Q1..Q4)
# df_y: columns 'Period', 'Total_Reports', 'SIF_Potential' (Yearly: 2021..2025)

def load_data():
    months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
    total_m = [120, 155, 235, 210, 285, 255, 310, 440, 310, 335, 425, 335]
    sif_m   = [42, 63, 94, 82, 115, 112, 138, 218, 140, 128, 195, 142]
    
    df_m = pd.DataFrame({'Period': months, 'Total_Reports': total_m, 'SIF_Potential': sif_m})
    
    # quarterly automatic calculate ho raha hai monthly se
    q_labels = ['Q1', 'Q2', 'Q3', 'Q4']
    total_q = [sum(total_m[0:3]), sum(total_m[3:6]), sum(total_m[6:9]), sum(total_m[9:12])]
    sif_q   = [sum(sif_m[0:3]), sum(sif_m[3:6]), sum(sif_m[6:9]), sum(sif_m[9:12])]
    df_q = pd.DataFrame({'Period': q_labels, 'Total_Reports': total_q, 'SIF_Potential': sif_q})
    
    # yearly trend (last 5 years)
    y_labels = ['2021', '2022', '2023', '2024', '2025']
    total_y = [2450, 2780, 3120, 3450, sum(total_m)]
    sif_y   = [620, 710, 830, 940, sum(sif_m)]
    df_y = pd.DataFrame({'Period': y_labels, 'Total_Reports': total_y, 'SIF_Potential': sif_y})
    
    return df_m, df_q, df_y

df_m, df_q, df_y = load_data()
```

```python
fig = go.Figure()

c_blue = '#00B0FF'
c_red = '#FF1744'

m_x = df_m['Period'].tolist()
m_tot = df_m['Total_Reports'].tolist()
m_sif = df_m['SIF_Potential'].tolist()

q_x = df_q['Period'].tolist()
q_tot = df_q['Total_Reports'].tolist()
q_sif = df_q['SIF_Potential'].tolist()

y_x = df_y['Period'].tolist()
y_tot = df_y['Total_Reports'].tolist()
y_sif = df_y['SIF_Potential'].tolist()

fig.add_trace(go.Scatter(
    x=m_x, y=m_tot, mode='lines',
    line=dict(color='rgba(0, 176, 255, 0.25)', width=8, shape='spline', smoothing=1.1),
    hoverinfo='skip', showlegend=False
))

fig.add_trace(go.Scatter(
    x=m_x, y=m_sif, mode='lines',
    line=dict(color='rgba(255, 23, 68, 0.25)', width=8, shape='spline', smoothing=1.1),
    hoverinfo='skip', showlegend=False
))

fig.add_trace(go.Scatter(
    x=m_x, y=m_tot, name='Total Reports', mode='lines+markers',
    line=dict(color=c_blue, width=3, shape='spline', smoothing=1.1),
    marker=dict(size=10, color='#000', symbol='circle', line=dict(color=c_blue, width=2.5)),
    fill='tozeroy', fillcolor='rgba(0, 176, 255, 0.08)',
    cliponaxis=False,
    hovertemplate='<b>Total Reports</b>: %{y:,}<extra></extra>'
))

fig.add_trace(go.Scatter(
    x=m_x, y=m_sif, name='SIF-Potential', mode='lines+markers',
    line=dict(color=c_red, width=3, shape='spline', smoothing=1.1),
    marker=dict(size=10, color='#000', symbol='circle', line=dict(color=c_red, width=2.5)),
    fill='tozeroy', fillcolor='rgba(255, 23, 68, 0.12)',
    cliponaxis=False,
    hovertemplate='<b>SIF-Potential</b>: %{y:,}<extra></extra>'
))

fig.update_layout(
    title=dict(
        text='SIF Reports Trend',
        font=dict(color='#FFFFFF', size=22, family='Inter, sans-serif'),
        x=0.03, y=0.94
    ),
    plot_bgcolor='#000000',
    paper_bgcolor='#000000',
    font=dict(family='Inter, sans-serif', color='#8B949E'),
    margin=dict(l=75, r=50, t=100, b=55),
    legend=dict(
        orientation='h',
        yanchor='middle',
        y=0.94,
        xanchor='center',
        x=0.55,
        font=dict(size=13, color='#E0E6ED'),
        bgcolor='rgba(0,0,0,0)',
        itemclick=False,
        itemdoubleclick=False
    ),
    updatemenus=[dict(
        buttons=list([
            dict(
                label='Monthly', method='update',
                args=[
                    {'x': [m_x, m_x, m_x, m_x], 'y': [m_tot, m_sif, m_tot, m_sif]},
                    {'xaxis.range': [-0.6, len(m_x) - 0.4], 'xaxis.tickvals': m_x, 'yaxis.autorange': True}
                ]
            ),
            dict(
                label='Quarterly', method='update',
                args=[
                    {'x': [q_x, q_x, q_x, q_x], 'y': [q_tot, q_sif, q_tot, q_sif]},
                    {'xaxis.range': [-0.6, len(q_x) - 0.4], 'xaxis.tickvals': q_x, 'yaxis.autorange': True}
                ]
            ),
            dict(
                label='Yearly', method='update',
                args=[
                    {'x': [y_x, y_x, y_x, y_x], 'y': [y_tot, y_sif, y_tot, y_sif]},
                    {'xaxis.range': [-0.6, len(y_x) - 0.4], 'xaxis.tickvals': y_x, 'yaxis.autorange': True}
                ]
            )
        ]),
        direction='down',
        showactive=True,
        x=0.98,
        xanchor='right',
        y=0.96,
        yanchor='middle',
        bgcolor='#182035',
        bordercolor='rgba(100, 160, 255, 0.4)',
        font=dict(color='#FFFFFF', size=13)
    )],
    xaxis=dict(
        showgrid=True,
        gridcolor='rgba(255, 255, 255, 0.06)',
        gridwidth=1,
        showline=False,
        zeroline=False,
        tickfont=dict(size=13, color='#8B949E'),
        tickmode='array',
        tickvals=m_x,
        range=[-0.6, len(m_x) - 0.4],
        fixedrange=True
    ),
    yaxis=dict(
        showgrid=True,
        gridcolor='rgba(255, 255, 255, 0.06)',
        gridwidth=1,
        showline=False,
        zeroline=False,
        tickfont=dict(size=13, color='#8B949E'),
        title=dict(text='Number of Reports', font=dict(size=15, color='#8B949E'), standoff=15),
        fixedrange=True
    ),
    hovermode='x unified',
    hoverlabel=dict(
        bgcolor='rgba(10, 14, 30, 0.95)',
        font_size=14,
        font_family='Inter, sans-serif',
        bordercolor='rgba(100, 160, 255, 0.5)'
    )
)

html_file = 'sif_reports_trend.html'
fig.write_html(html_file, include_plotlyjs='cdn', config={'displayModeBar': False})

with open(html_file, 'r', encoding='utf-8') as f:
    html_content = f.read()

custom_js = '''
<style>
.updatemenu-button:hover text.updatemenu-item-text {
    fill: #000000 !important;
}
</style>
<script>
function syncDropdownText() {
    document.querySelectorAll('.updatemenu-button, .updatemenu-dropdown-button').forEach(function(btn) {
        var rect = btn.querySelector('rect.updatemenu-item-rect');
        var text = btn.querySelector('text.updatemenu-item-text');
        if (rect && text) {
            var fill = window.getComputedStyle(rect).fill;
            var m = fill.match(/\\d+/g);
            if (m && m.length >= 3) {
                var lum = (parseInt(m[0])*299 + parseInt(m[1])*587 + parseInt(m[2])*114) / 1000;
                if (lum > 120) {
                    text.style.setProperty('fill', '#000000', 'important');
                } else {
                    text.style.setProperty('fill', '#FFFFFF', 'important');
                }
            }
        }
    });
}
new MutationObserver(syncDropdownText).observe(document.body, { childList: true, subtree: true, attributes: true, attributeFilter: ['style', 'fill'] });
document.addEventListener('mousemove', syncDropdownText);
document.addEventListener('click', syncDropdownText);
setInterval(syncDropdownText, 100);
</script>
'''

html_content = html_content.replace('</body>', custom_js + '</body>')
with open(html_file, 'w', encoding='utf-8') as f:
    f.write(html_content)

display(IFrame(html_file, width="100%", height=550))
```
